```python
import pandas as pd
import re
```

```python
#load data
df = pd.read_csv("final_data.csv")
df = df[['Review', 'Rating']]
```

```python
#view
df.head()
```

unique words

```python
# Function to clean and tokenize text
def tokenize(text):
    text = str(text).lower()                      # lowercase
    text = re.sub(r'[^a-z\s]', '', text)          # keep only letters & spaces
    words = text.split()
    return set(words)                             # return unique words in this review

# Dictionary to store unique words per rating
rating_words = {}

for rating in df['Rating'].unique():
    # Filter reviews for this rating
    reviews = df[df['Rating'] == rating]['Review']
    
    # Collect all unique words across reviews of this rating
    unique_words = set()
    for review in reviews:
        unique_words.update(tokenize(review))
    
    rating_words[rating] = unique_words

# Print results
for rating, words in rating_words.items():
    print(f"\nRating {rating} has {len(words)} unique words:")
    print(list(words)[:30])  # show first 30 words as sample
```

unique & exclussive words

```python
# Step 1: Collect all words per rating
rating_words = {}
for rating in df['Rating'].unique():
    reviews = df[df['Rating'] == rating]['Review']
    unique_words = set()
    for review in reviews:
        unique_words.update(tokenize(review))
    rating_words[rating] = unique_words

# Step 2: Find words that are unique ONLY to each rating
exclusive_words = {}
all_ratings = set(rating_words.keys())

for rating in all_ratings:
    # words in this rating
    words_in_rating = rating_words[rating]
    # words in all other ratings
    words_in_others = set().union(*[rating_words[r] for r in all_ratings if r != rating])
    # exclusive words = present only in this rating
    exclusive_words[rating] = words_in_rating - words_in_others

# ------------------------
# Results
# ------------------------
for rating in rating_words:
    print(f"\nRating {rating}:")
    print(f"  Total unique words = {len(rating_words[rating])}")
    print(f"  Exclusive words    = {len(exclusive_words[rating])}")
    print(f"  Sample exclusive words: {list(exclusive_words[rating])[:30]}")
```

```python
# Step 3: Sentiment filtering from exclusive words
# ------------------------
import nltk
from nltk.corpus import opinion_lexicon

# Download lexicon if not already installed
nltk.download('opinion_lexicon')

# Positive and negative lexicon
positive_words = set(opinion_lexicon.positive())
negative_words = set(opinion_lexicon.negative())

exclusive_sentiment_words = {}

for rating in exclusive_words:
    pos_matches = exclusive_words[rating].intersection(positive_words)
    neg_matches = exclusive_words[rating].intersection(negative_words)
    
    exclusive_sentiment_words[rating] = {
        "positive": list(pos_matches),
        "negative": list(neg_matches)
    }

# ------------------------
# Results
# ------------------------
for rating in exclusive_sentiment_words:
    print(f"\nRating {rating} (exclusive sentiment words):")
    print(f"  Positive: {exclusive_sentiment_words[rating]['positive']}")
    print(f"  Negative: {exclusive_sentiment_words[rating]['negative']}")
```

most repeated words

```python
from collections import Counter

# Find most repeated words per rating
most_common_words = {}

for rating in df['Rating'].unique():
    reviews = df[df['Rating'] == rating]['Review']
    all_words = []
    for review in reviews:
        all_words.extend(tokenize(review))
    
    counter = Counter(all_words)
    most_common_words[rating] = counter.most_common(20)  # top 20 words
    
# Show results
for rating, words in most_common_words.items():
    print(f"\nRating {rating} - Top repeated words:")
    for word, count in words:
        print(f"{word}: {count}")
```

list of products

```python
# List of fashion product keywords (expand this as per your dataset)
product_keywords = [
    "shirt", "jeans", "bag", "shoes", "jacket","glasses",
    "saree", "kurti", "tshirt", "sandals", "skirt", "top",
    "trousers", "hoodie", "blouse", "leggings", "shorts"
]

# Function to extract mentioned products
def extract_products(review):
    review = str(review).lower()
    found = [prod for prod in product_keywords if prod in review]
    return found if found else None

# Apply function to dataset
df["Products_Mentioned"] = df["Review"].apply(extract_products)

# Flatten product mentions and count frequencies
all_products = []
for prods in df["Products_Mentioned"].dropna():
    all_products.extend(prods)

product_counts = Counter(all_products)

# Convert to DataFrame for better view
product_freq_df = pd.DataFrame(product_counts.items(), columns=["Product", "Review_Count"])
product_freq_df = product_freq_df.sort_values(by="Review_Count", ascending=False)

print(product_freq_df)
```
